```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from memoranda.paths import MANIFESTS, TABLES, FIGURES
from memoranda.analysis.tables import image_table, subject_image_table
pd.set_option('display.width', 160)
```

## The pictures

```python
it = image_table()
print(len(it), 'unique images')
it.category.value_counts()
```

```python
from IPython.display import Image
Image(FIGURES / 'montage_all_unique.png', width=1400)
```

## Units and concept cells

```python
sel = pd.read_csv(MANIFESTS / 'unit_selectivity.csv')
sel.groupby(['task','area']).concept_cell.agg(['sum','count','mean']).round(3)
```

## Memoranda vs the rest of each subject's screening set

```python
st = subject_image_table()
scr = st[st.in_screening]
scr.groupby('in_sternberg')[['n_concept_pref','n_concept_pref_mtl','z_max_mtl','face_found']].mean().round(3)
```

## Re-derive one number by hand: per-image concept-cell rate by category

```python
score = pd.read_csv(TABLES / 'A5_image_neural_score.csv')
score[score.n_subjects>=3].groupby('category').concept_per_subject.agg(['mean','count']).sort_values('mean', ascending=False).round(3)
```
